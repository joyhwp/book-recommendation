## 데이터 정리하기

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("dylanjcastillo/7k-books-with-metadata")

print("Path to dataset files:", path)
```

```python
import pandas as pd
```

```python
books=pd.read_csv(f"{path}/books.csv")
```

```python
books
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
ax=plt.axes()
sns.heatmap(books.isna().transpose(), cbar=False, ax=ax)
plt.xlabel("Columns")
plt.ylabel("missing")
```

- subtitles are mostly missing
- there are some descriptions missing-> bias없는지 확인해야함.

```python
import numpy as np
books["missing_description"]=np.where(books["description"].isna(),1,0)
books["age_of_book"]=2024-books["published_year"]
```

```python
columns_of_interests=["num_pages","age_of_book","missing_description","average_rating"]
correlation_matrix=books[columns_of_interests].corr(method="spearman")#noncontinous변수 다룰때 더 용이
sns.set_theme(style="white")
plt.figure(figsize=(8,6))
heatmap=sns.heatmap(correlation_matrix,annot=True, fmt=".2f",cmap="coolwarm",cbar_kws={"label":"spearman correlation"})
heatmap.set_title("correlation heatmap")
plt.show()
```

no high corr-> bias 없음/제거해도됨

```python
book_missing=books[~(books["description"].isna())&
      ~(books["num_pages"].isna())&
      ~(books["average_rating"].isna())&
      ~(books["published_year"].isna())]
```

```python
book_missing
```

고유한 categories가 너무 많음 (531)

```python
book_missing["categories"].value_counts().reset_index().sort_values("count",ascending=False)
```

매우 편향됨
description도 유용한 정보 담고 있는지 확인해야함.

```python
book_missing["words_in_description"]=book_missing["description"].str.split().str.len()
```

```python
book_missing
```

cut point 지정해야함.

```python
book_missing.loc[book_missing["words_in_description"].between(1,4),"description"]
```

```python
book_missing.loc[book_missing["words_in_description"].between(5,14),"description"]
```

```python
book_missing.loc[book_missing["words_in_description"].between(15,24),"description"]
```

```python
book_missing.loc[book_missing["words_in_description"].between(25,34),"description"]
```

```python
book_missing_25=book_missing[book_missing["words_in_description"]>=25]
```

```python
book_missing_25
```

subtitle 어떻게 할까-> title와 pair하기

```python
book_missing_25["title_subtitle"] = book_missing_25.apply(
    lambda row: str(row["title"]) if pd.isna(row["subtitle"])
    else f'{row["title"]}: {row["subtitle"]}',
    axis=1
)
```

```python
book_missing_25
```

description과 isbn 고유값 이어주기

```python
book_missing_25["tagged_description"]=book_missing_25[["isbn13","description"]].astype(str).agg(" ".join,axis=1)
```

```python
book_missing_25
```

```python
(
    book_missing_25
    .drop(["subtitle","missing_description","age_of_book","words_in_description"],axis=1)
    .to_csv("books_cleaned.csv",index=False)
)
```
